Same Data, Two Aspect Ratios

Aesthetic: both are clean. Substantive: both are accurate — no data is changed. Perceptual: the tall narrow version makes the growth look dramatically steeper; the wide version makes it look more gradual. Demonstrates that aspect ratio alone can change the story a visualization tells.

```python
# ============================================================
# SESSION 2: Getting Started with Matplotlib
# ============================================================

# --- Import Libraries ---
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
from PIL import Image
import requests

# ============================================================
# MAKING A BASIC FIGURE
# ============================================================

# Generate sample data
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50)

# --- Scatter plot ---
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y)
plt.show()

# --- Bar chart ---
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x, y)
plt.show()

# --- Line plot ---
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
plt.show()

# ============================================================
# ACTIVITY: Produce a histogram
# ============================================================
# Note: hist only takes one variable (y), not x and y
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
plt.show()

# ============================================================
# PLOT LABELS
# ============================================================

# Adding axis labels and title
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
plt.show()

# --- Font dictionaries ---
font1 = {'family': 'sans-serif', 'color': 'blue', 'size': 20}
font2 = {'family': 'monospace', 'color': 'green', 'size': 14}

# Applying font dictionaries to labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
ax.set_title('Total growth over time', fontdict=font1)
ax.set_ylabel('Total growth', fontdict=font2)
ax.set_xlabel('Years since start', fontdict=font2)
fig.tight_layout()
plt.show()

# Moving labels with loc argument
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
ax.set_ylabel('Total growth', fontdict=font2, loc='top')
ax.set_xlabel('Years since start', fontdict=font2, loc='right')
fig.tight_layout()
plt.show()

# ============================================================
# CUSTOMIZING PLOT APPEARANCE
# ============================================================

# Scatter with color and marker
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,
           y,
           marker='*',
           color='indigo')
plt.show()

# Line with linestyle and linewidth
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,
        y,
        marker='*',
        color='indigo',
        linestyle='--',
        linewidth=2)
plt.show()

# Using hex color codes
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='*', color='#7425b9', linestyle='--', linewidth=2)
plt.show()

# Full marker customization
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y,
        marker='*',
        markersize=12,
        color='#7425b9',
        linestyle='--',
        linewidth=2,
        markeredgecolor='#fa9359',
        markerfacecolor='#000000')
fig.tight_layout()
plt.show()

# ============================================================
# GRID LINES
# ============================================================

# Basic grid on y-axis
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='*', markersize=12, color='#7425b9',
        linestyle='--', linewidth=2,
        markeredgecolor='#fa9359', markerfacecolor='#000000')
ax.grid(axis='y')
fig.tight_layout()
plt.show()

# ============================================================
# ACTIVITY: Modify grid line appearance
# ============================================================
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='*', markersize=12, color='#7425b9',
        linestyle='--', linewidth=2,
        markeredgecolor='#fa9359', markerfacecolor='#000000')
ax.grid(axis='y', color='blue', linewidth=2, linestyle='-.')
fig.tight_layout()
plt.show()

# ============================================================
# ACTIVITY: Exploring Matplotlib — Python Graph Gallery
# Chosen chart type: Bubble Chart
# Source: https://python-graph-gallery.com/bubble-plot/
# ============================================================

# A bubble chart extends a scatter plot by adding a third
# variable encoded as the size of each point.

np.random.seed(613)
n = 30
x_bubble = np.random.rand(n) * 100
y_bubble = np.random.rand(n) * 100
size = np.random.rand(n) * 1000   # bubble size
colors = np.random.rand(n)        # color variation

fig, ax = plt.subplots(figsize=(7, 5))
scatter = ax.scatter(x_bubble, y_bubble,
                     s=size,
                     c=colors,
                     cmap='viridis',
                     alpha=0.6,
                     edgecolors='black',
                     linewidths=0.5)
ax.set_title('Bubble Chart Example', fontsize=14)
ax.set_xlabel('X Value')
ax.set_ylabel('Y Value')
fig.colorbar(scatter, ax=ax, label='Color Scale')
fig.tight_layout()
plt.show()

# Evaluation:
# Aesthetic: clean, uses color and size effectively to add visual interest
#            without cluttering the plot.
# Substantive: accurately represents three variables simultaneously
#              (x position, y position, and magnitude via bubble size).
#              The use of both size and color for separate variables
#              adds information density without distorting any single variable.
# Perceptual: the message (relative magnitude of data points across
#             two dimensions) is immediately readable. However, human
#             perception of area is less precise than length, so bubble
#             charts are better for approximate comparisons than exact ones.
```
